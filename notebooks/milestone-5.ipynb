Q1: Dataset Splitting Strategy

```python
from sklearn.model_selection import train_test_split

genres = 10
recipes_per_genre = 100
total_recipes = genres * recipes_per_genre

recipes = list(range(total_recipes))

train_recipes, val_recipes = train_test_split(recipes, test_size=0.2, shuffle=True, random_state=42)

q1_ans = len(val_recipes)
q1_ans
```

Q2: On-the-Fly Mixing Dimensions

```python
import numpy as np

stem1 = np.zeros(160000)
stem2 = np.zeros(160000)
stem3 = np.zeros(160000)
stem4 = np.zeros(160000)
noise = np.zeros(160000)

mix = stem1 + stem2 + stem3 + stem4
mix = mix + (noise * 0.2)

q2_ans = mix.shape
q2_ans
```

Q3: Hugging Face Feature Extractor Shape

```python
from transformers import AutoFeatureExtractor
import numpy as np

extractor = AutoFeatureExtractor.from_pretrained("MIT/ast-finetuned-audioset-10-10-0.4593")

mix = np.ones(160000)

extracted = extractor(mix, sampling_rate=16000, return_tensors="pt")
input_values = extracted["input_values"]
squeezed_tensor = input_values.squeeze(0)

q3_ans = squeezed_tensor.shape
list(q3_ans)
```

Q4: Model Architecture Initialization

```python
from transformers import ASTForAudioClassification

model = ASTForAudioClassification.from_pretrained("MIT/ast-finetuned-audioset-10-10-0.4593", num_labels=10, ignore_mismatched_sizes=True)

trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

q4_ans = trainable_params
q4_ans
```

Q5: Inference Normalization Math

```python
import numpy as np

y_test = np.array([-0.85, 0.40, 0.20, -0.10])
y_normalized = y_test / (np.max(np.abs(y_test)) + 1e-9)

q5_ans = round(y_normalized[0], 3)
q5_ans
```
